Intial chunk split (Naive)

```python
def _split_initially(text, chunk_size, overlap):
        """Split text into sentence-aware chunks with word overlap."""
        text = re.sub(r"\s+", " ", text).strip()
        if not text:
            return []

        sentences = re.split(r"(?<=[.!?])\s+", text)
        chunks = []
        current_words = []

        for sentence in sentences:
            sentence_words = sentence.split()
            if not sentence_words:
                continue

            # Handle sentences longer than the chunk limit.
            if len(sentence_words) > chunk_size:
                if current_words:
                    chunks.append(" ".join(current_words))
                    current_words = []

                step = chunk_size - overlap
                for i in range(0, len(sentence_words), step):
                    part = sentence_words[i:i + chunk_size]
                    if part:
                        chunks.append(" ".join(part))
                continue

            if len(current_words) + len(sentence_words) <= chunk_size:
                current_words.extend(sentence_words)
            else:
                if current_words:
                    chunks.append(" ".join(current_words))

                # Preserve overlap from the previous chunk.
                overlap_words = current_words[-overlap:] if overlap else []
                current_words = overlap_words + sentence_words

                # If overlap plus sentence exceeds the limit, trim overlap.
                if len(current_words) > chunk_size:
                    current_words = current_words[-chunk_size:]

        if current_words:
            chunks.append(" ".join(current_words))

        # Remove empty chunks.
        return [chunk for chunk in chunks if chunk.strip()]
```

Embedding and Indexing Chunks

```python
import torch
import pyarrow as pa
from tqdm.auto import tqdm
from sentence_transformers import SentenceTransformer

# 1. Load embedder directly onto GPU
embed_model = SentenceTransformer("BAAI/bge-small-en-v1.5", device="cuda" if torch.cuda.is_available() else "cpu")

def fast_ingest_wiki(wiki_source, table, limit=1000, chunk_size=500, overlap=20, batch_size=256):
    print(f"Streaming and chunking {limit} documents...")
    all_records = []

    # Fast in-memory chunking (CPU string ops are instant for 1000 docs)
    iterator = iter(wiki_source)
    for i in tqdm(range(limit), desc="1/3 Chunking Docs"):
        try:
            doc = next(iterator)
        except StopIteration:
            break

        doc_id = str(doc.get("id", doc.get("title", f"doc_{i}")))
        text = doc.get("text", "")
        if not text:
            continue

        chunks = _split_initially(text, chunk_size=chunk_size, overlap=overlap)
        if not chunks:
            chunks = [text]

        for c_idx, chunk in enumerate(chunks):
            all_records.append({
                "doc_id": doc_id,
                "chunk_id": c_idx,
                "text": chunk
            })

    total_chunks = len(all_records)
    print(f"Generated {total_chunks} chunks. Computing GPU embeddings...")

    # 2. Batch embed on GPU with progress bar
    texts = [r["text"] for r in all_records]
    embeddings = embed_model.encode(
        texts,
        batch_size=batch_size,
        show_progress_bar=True,
        normalize_embeddings=True,
        convert_to_numpy=True
    )

    # Attach vectors
    for idx, r in enumerate(all_records):
        r["vector"] = embeddings[idx]

    print("Writing to LanceDB in batches...")
    # 3. Add to table in chunks
    for i in tqdm(range(0, total_chunks, 5000), desc="3/3 Writing to DB"):
        batch = all_records[i:i + 5000]
        table.add(batch)

    print("Optimizing LanceDB & building FTS index...")
    table.optimize()
    table.create_fts_index("text", replace=True)
    print(f"Done! Final rows: {table.count_rows()}")
```

Query Intent Adaptive Retrieval

```python
import json
import re
import torch

class QueryIntentAdaptiveRetrieval:
    def __init__(self, model, tokenizer, table):
        self.model = model
        self.tokenizer = tokenizer
        self.table = table
        self.last_weights = None
        self.last_enriched = None

    def _generate(self, messages, max_new_tokens=60):
        inputs = self.tokenizer.apply_chat_template(
            messages,
            add_generation_prompt=True,
            tokenize=True,
            return_tensors="pt",
            return_dict=True
        )
        device = self.model.get_input_embeddings().weight.device
        inputs = {k: v.to(device) for k, v in inputs.items()}
        input_len = inputs["input_ids"].shape[-1]

        with torch.no_grad():
            outputs = self.model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                do_sample=False,
                pad_token_id=self.tokenizer.eos_token_id
            )
        return self.tokenizer.decode(outputs[0][input_len:], skip_special_tokens=True).strip()

    def query_enrichment(self, query):
        pd = self._generate([
            {"role": "system", "content": "Paraphrase this query into a semantic statement for dense retrieval. Output only the statement."},
            {"role": "user", "content": query}
        ], max_new_tokens=60)

        ps = self._generate([
            {"role": "system", "content": "Extract only the most critical, search-relevant keywords, unique nouns, and core entities from this query. Format: Return ONLY a comma-separated list of the 3 to 6 key terms. No conversational words, no introductory text."},
            {"role": "user", "content": query}
        ], max_new_tokens=40)

        pe = self._generate([
            {"role": "system", "content": "Write a 1-sentence hypothetical passage answering this query. Output only the passage."},
            {"role": "user", "content": query}
        ], max_new_tokens=60)

        return {"pd": pd, "ps": ps, "pe": pe}

    def dynamic_channel_weighting(self, query: str) -> list[float]:
        prompt = (
            "Assign 3 weights [dense, sparse, extended] summing to 1.0 based on search intent:\n"
            "- dense: high for conceptual, broad, or semantic meaning\n"
            "- sparse: high for exact entities, IDs, names, dates, quotes\n"
            "- extended: high for complex or exploratory QA requiring synthesis\n\n"
            f"Query: {query}\n"
            "Output ONLY a valid JSON array of 3 numbers, e.g. [0.5, 0.3, 0.2]. No explanation."
        )
        try:
            # 1. Bump token limit so formatting tokens don't truncate the payload
            raw = self._generate(
                [
                    {
                        "role": "system",
                        "content": (
                            "You are a strict JSON generator. Return only a JSON array of 3 floats."
                        ),
                    },
                    {"role": "user", "content": prompt},
                ],
                max_new_tokens=60,
            )

            # 2. Extract strictly valid decimal/integer floats using regex
            # Matches patterns like 0.4, .4, 1.0, 1
            float_matches = re.findall(r"(?:\d+\.?\d*|\.\d+)", raw)

            if len(float_matches) >= 3:
                w = [float(f) for f in float_matches[:3]]
            else:
                # If regex doesn't catch 3, try JSON array extraction
                json_match = re.search(r"\[[\d\s.,]+\]", raw)
                if json_match:
                    w = json.loads(json_match.group(0))
                else:
                    w = [0.4, 0.3, 0.3]  # Intentional default, not uniform

            total = sum(w)
            if total <= 0:
                return [0.34, 0.33, 0.33]

            # Normalize and round cleanly
            return [round(x / total, 3) for x in w]

        except Exception as e:
            # Print or log the error during development so failures aren't hidden
            print(f"[Warning] Weighting fallback triggered: {e} | Raw output: {raw!r}")
            return [0.34, 0.33, 0.33]

    def retrieve(self, query, k=3):
        enriched = self.query_enrichment(query)
        weights = self.dynamic_channel_weighting(query)
        self.last_enriched, self.last_weights = enriched, weights
        alpha, beta, gamma = weights

        fetch_limit = k * 4
        dense_scores, sparse_scores, extended_scores = {}, {}, {}
        meta_lookup = {}

        def record_meta(item):
            t = item["text"]
            if t not in meta_lookup:
                meta_lookup[t] = {
                    "text": t,
                    "doc_id": item.get("doc_id", "unknown"),
                    "chunk_id": item.get("chunk_id", 0)
                }

        # 1. Dense Semantic (pd)
        try:
            for item in self.table.search(enriched["pd"], query_type="vector").metric("cosine").limit(fetch_limit).to_list():
                t = item["text"]
                record_meta(item)
                dense_scores[t] = max(dense_scores.get(t, -1.0), 1.0 - item.get("_distance", 0.0))
        except Exception as e:
            print(f"Dense search error: {e}")

        # 2. Sparse Lexical (ps)
        try:
            for item in self.table.search(enriched["ps"], query_type="fts").limit(fetch_limit).to_list():
                t = item["text"]
                record_meta(item)
                sparse_scores[t] = max(sparse_scores.get(t, 0.0), item.get("_score", 0.0))
        except Exception as e:
            print(f"Sparse search error: {e}")

        # 3. Extended Context (pe)
        try:
            for item in self.table.search(enriched["pe"], query_type="vector").metric("cosine").limit(fetch_limit).to_list():
                t = item["text"]
                record_meta(item)
                extended_scores[t] = max(extended_scores.get(t, -1.0), 1.0 - item.get("_distance", 0.0))
        except Exception as e:
            print(f"Extended search error: {e}")

        all_texts = set(dense_scores) | set(sparse_scores) | set(extended_scores)
        if not all_texts:
            return []

        def norm(scores):
            if not scores:
                return {}
            vals = scores.values()
            mn, mx = min(vals), max(vals)
            denom = (mx - mn) if mx != mn else 1.0
            return {k: (v - mn) / denom for k, v in scores.items()}

        nd, ns, ne = norm(dense_scores), norm(sparse_scores), norm(extended_scores)

        ranked = []
        for t in all_texts:
            score = (alpha * nd.get(t, 0.0)) + (beta * ns.get(t, 0.0)) + (gamma * ne.get(t, 0.0))
            entry = meta_lookup[t].copy()
            entry["score"] = score
            ranked.append(entry)

        ranked.sort(key=lambda x: x["score"], reverse=True)
        return ranked[:k]
```

Semantic-Preserving Chunking Functions

```python
import torch
import re
import json

class SemanticSLMChunker:
    def __init__(self, model, tokenizer, chunk_size=500, overlap=100):
        self.model = model
        self.tokenizer = tokenizer
        self.chunk_size = chunk_size
        self.overlap = overlap

        if self.chunk_size <= 0:
            raise ValueError("chunk_size must be positive.")
        if not 0 <= self.overlap < self.chunk_size:
            raise ValueError("overlap must be >= 0 and < chunk_size.")

    def _generate(self, messages, max_new_tokens=100):
        """Generate text using the SLM """
        inputs = self.tokenizer.apply_chat_template(
            messages,
            add_generation_prompt=True,
            tokenize=True,
            return_tensors="pt",
            return_dict=True
        )
        input_device = self.model.get_input_embeddings().weight.device
        inputs = {key: value.to(input_device) for key, value in inputs.items()}
        input_length = inputs["input_ids"].shape[-1]

        with torch.no_grad():
            outputs = self.model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                do_sample=False,
                pad_token_id=self.tokenizer.eos_token_id
            )
        return self.tokenizer.decode(outputs[0][input_length:], skip_special_tokens=True).strip()

    def detect_damage(self, chunk, query=""):
        """Determine whether this text chunk has critical truncation or syntax issues."""
        prompt = (
            "Determine whether this text chunk has a critical truncation, "
            "broken syntax, or an unresolved reference caused by a cutoff. "
            "A chunk is not damaged merely because it starts or ends mid-topic.\n\n"
            f"Query Context: {query}\n\n"
            f"Text chunk:\n{chunk}\n\n"
            "Respond with exactly one word: DAMAGED or INTACT."
        )
        try:
            answer = self._generate([
                {"role": "system", "content": "You are a precise linguistic evaluator. Output only DAMAGED or INTACT."},
                {"role": "user", "content": prompt}
            ], max_new_tokens=10).upper()
            return "DAMAGED" in answer
        except Exception:
            return False

    def compress_stitch(self, pre, target, post, query=""):
        """Repair the TARGET chunk using previous and next segments as context."""
        prompt = (
            "Repair the TARGET chunk only, using the previous and next segments as context. "
            "Preserve original meaning, entities, and technical details. "
            "Fix cutoffs and broken transitions. Return only the repaired target text.\n\n"
            f"Query: {query}\n\n"
            f"PREVIOUS SEGMENT:\n{pre}\n\n"
            f"TARGET SEGMENT:\n{target}\n\n"
            f"NEXT SEGMENT:\n{post}\n\n"
            "Repaired target:"
        )
        try:
            return self._generate([
                {"role": "system", "content": "You are a careful text-repair assistant. Preserve context and do not fabricate details."},
                {"role": "user", "content": prompt}
            ], max_new_tokens=300)
        except Exception:
            return target

    def extract_key_information_elements(self, query):
        """Extract key fact elements from query to audit retrieval completeness with formatting cleanup."""
        prompt = (
            f"List the 2-4 core factual requirements needed to answer this query.\n"
            f"Query: {query}\n"
            f"Format: Return ONLY a comma-separated list of short requirements."
        )
        try:
            raw = self._generate([
                {"role": "system", "content": "Output only a comma-separated list of requirements."},
                {"role": "user", "content": prompt}
            ], max_new_tokens=100)

            # Apply cleaning logic to strip prefixes, headers and numbers
            items = []
            lines = re.split(r'[\n,;]+', raw)
            for line in lines:
                cleaned = re.sub(r'^\s*\d+[:\.)]\s*|^\s*[-*+]\s*', '', line).strip()
                if cleaned and len(cleaned) > 3 and not any(header in cleaned.lower() for header in ["factual requirement", "here are", "core requirement", "to answer"]):
                    items.append(cleaned)

            return items if items else [query]
        except Exception:
            return [query]

    def audit_coverage(self, evidence_chunks, key_elements):
        """Assess coverage/entailment of extracted elements against gathered evidence."""
        context_text = "\n\n".join(evidence_chunks)[:2500]
        missing = []
        for element in key_elements:
            prompt = (
                f"Context:\n{context_text}\n\n"
                f"Does the context above explicitly contain information about: '{element}'?\n"
                f"Answer ONLY 'YES' or 'NO'."
            )
            try:
                verdict = self._generate([
                    {"role": "system", "content": "You are a factual entailment verifier. Output only YES or NO."},
                    {"role": "user", "content": prompt}
                ], max_new_tokens=5).strip().upper()
                if "YES" not in verdict:
                    missing.append(element)
            except Exception:
                missing.append(element)
        return missing

# Instantiate the clean, structured semantic chunker
text_splitter = SemanticSLMChunker(
    model=model,
    tokenizer=tokenizer,
    chunk_size=500,
    overlap=100
)
```

Wrapped in the chain pipeline for easy use

```python
from langchain_core.runnables import RunnablePassthrough, RunnableLambda

def run_insem_rag_pipeline(inputs, max_iters=3):
    initial_query = inputs if isinstance(inputs, str) else inputs.get("query", "")
    current_query = initial_query

    accumulated_chunks = []
    seen_texts = set()

    for iteration in range(1, max_iters + 1):
        print(f"\n--- [ITERATION {iteration}/{max_iters}] Query: '{current_query}' ---")

        # 1. Intent-Adaptive Retrieval (IAR)
        retrieved_results = retriever_engine.retrieve(current_query, k=3)
        raw_chunks = [item["text"] for item in retrieved_results]
        print(f"Calculated Hybrid Weights: {retriever_engine.last_weights}")
        print(f"Enriched Semantic (pd): {retriever_engine.last_enriched['pd']}")
        print(f"Enriched Keywords (ps): {retriever_engine.last_enriched['ps']}\n")
        # 2. Semantic-Preserving Chunking (SPC): Repair damaged chunks
        repaired_chunks = []
        for idx, chunk in enumerate(raw_chunks):
            if text_splitter.detect_damage(chunk, query=current_query):
                pre = raw_chunks[idx - 1] if idx > 0 else ""
                post = raw_chunks[idx + 1] if idx < len(raw_chunks) - 1 else ""
                repaired = text_splitter.compress_stitch(pre, chunk, post, query=current_query)
                repaired_chunks.append(repaired)
            else:
                repaired_chunks.append(chunk)

        # Accumulate unique chunks
        for chunk in repaired_chunks:
            if chunk not in seen_texts:
                accumulated_chunks.append(chunk)
                seen_texts.add(chunk)

        # 3. Key Information Element (KIE) Coverage Audit against original query
        kie_elements = text_splitter.extract_key_information_elements(initial_query)
        missing_elements = text_splitter.audit_coverage(accumulated_chunks, kie_elements)

        # Check stopping condition
        if not missing_elements:
            print(f"  -> [Audit Success] All elements covered at iteration {iteration}.")
            break

        print(f"  -> [Audit Incomplete] Missing elements: {missing_elements}")

        # Reforumalte query for the next iteration
        current_query = f"{initial_query} focusing specifically on: {', '.join(str(e) for e in missing_elements)}"

    final_context = "\n\n".join(accumulated_chunks)
    return {"context": final_context, "query": initial_query}
```

Running full pipeline with generator(using the same SLM for now)

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableLambda

prompt = PromptTemplate.from_template(
    """Answer the question based only on the following context. If the context does not contain enough information to answer the question, state: "The provided context does not contain sufficient information to answer this question."\n\nContext:\n{context}\n\nQuestion: {query}\nAnswer:"""
)

def generate_answer(inputs):
    prompt_str = inputs.to_string()
    encoded = tokenizer(prompt_str, return_tensors="pt").to(model.device)
    input_length = encoded["input_ids"].shape[-1]

    with torch.no_grad():
        outputs = model.generate(
            **encoded,
            max_new_tokens=200,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )
    return tokenizer.decode(outputs[0][input_length:], skip_special_tokens=True).strip()

# 2. Build the chain
insem_rag_chain = (
    RunnableLambda(run_insem_rag_pipeline)
    | prompt
    | RunnableLambda(generate_answer)
)
```
